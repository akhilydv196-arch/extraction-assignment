**3.Explain the difference between CSV and Excel in terms of extraction and ETL usage**

-CSV (Comma-Separated Values) and Excel (XLSX) differ significantly in ETL (Extract, Transform, Load) processes because CSV is designed as a raw, flat data storage format meant for machines, while Excel is a binary format designed for human analysis, formatting, and presentation. In ETL, CSV is favored for speed and universality, while Excel is chosen when data requires manual pre-processing or retains complex formatting.

**4.Explain the steps involved in extracting data from a relational database.**

- Extracting data from a relational database involves establishing a connection, identifying required data, executing SQL queries, and exporting the results. Common techniques include full extraction, incremental extraction using timestamps/CDC (change data capture),

Steps to Extract Data from a Relational Database:


Identify Data Sources & Connect: Define the source tables and establish a connection to the database (e.g., MySQL, PostgreSQL, SQL Server) using JDBC/ODBC drivers, database clients, or languages like Python.


Define Extraction Method: Determine if a Full Extraction (all data) or Incremental Extraction (only new/updated data using timestamps or Change Data Capture/CDC) is required.


Create and Execute SQL Queries: Write SQL queries (using SELECT, WHERE, JOIN) to isolate specific data needed from tables.  


Extract/Export Data: Run the query to pull the data. Data is typically exported into a flat file format (e.g., CSV, JSON) or directly into a staging area.


Validate and Store Data: Verify the data for accuracy, completeness

 **6.What are APIs? Explain how APIs help in real-time data extraction.**

 An Application Programming Interface (API) is a set of defined rules and protocols that allow different software applications to communicate and share data with each other. It acts as an intermediary or "messenger," taking a request from a client application (like a mobile app) to a server and then delivering the server’s response back to the client.

 In traditional data extraction, systems often rely on "batch processing," where data is collected and updated at fixed intervals (e.g., daily or weekly). APIs enable real-time data extraction by allowing for instant or continuous exchange of information as soon as it is generated.

 **Immediate Request-Response**: Synchronous APIs (like REST APIs) allow applications to fetch the latest available data on-demand. For example, a stock trading platform uses APIs to pull up-to-the-second price updates the moment a user requests them.


**Event-Driven Push Mechanisms**: Instead of waiting for a request, some APIs use "push" technology to send data immediately when a specific event occurs. For instance, Webhooks automatically trigger a data transfer to your system the second a credit card is swiped or a new message is received.


**Persistent Connections**: Specialized protocols like WebSockets maintain an open, bidirectional connection between the client and server. This is vital for high-frequency updates, such as live sports scores or GPS tracking in ride-sharing apps, where data must flow continuously without the delay of repeated handshakes.


**Continuous Data Streaming**: Streaming APIs deliver a constant flow of data points, allowing for "real-time analytics." This enables businesses to detect fraud in milliseconds by analyzing transaction streams as they arrive.


**Breaking Down Silos***: APIs provide a standardized way to access  APIs provide a standardized way to access data that might otherwise be trapped in separate, incompatible systems. By acting as a bridge, they ensure that various departments—from sales to inventory—can all see and extract the same live version of the truth simultaneously.

 **8.What steps should an ETL developer take when extracting data from large CSV files (1GB+)?**

 Extracting data from large CSV files (1GB+) requires specialized techniques to avoid memory bottlenecks and slow processing times. ETL developers should prioritize memory management, parallelization, and columnar storage formats to ensure efficient data ingestion.
